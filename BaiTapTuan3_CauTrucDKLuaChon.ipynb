W3A1
Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ 
NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi. Chú ý: kiểm tra chỉ xử
lý các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý

```python
n = int(input("nhap vao so nguyen duong co 3 chu so: "))
if 100<= n <=999:
    don_vi = n % 10
    hang_chuc = (n//10) % 10
    hang_tram = n//100
    so_dao_nguoc = don_vi * 100 + hang_chuc * 10 + hang_tram
    print("số đảo ngược của",n,"là:",so_dao_nguoc)
```

W3A3
Kiểm tra xem một số có phải là lũy thừa của 2 hay không 
Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 
hay không. 
Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0

```python
n = int(input("nhap n: "))
if n % 2 == 0 and n & (n-1) == 0:
    print("n là lũy thừa của 2")
else:
    print("n không là lũy thừa của 2")
```

tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in

```python
a, b, c = map(int, input().split())
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Tam giác đều")
    elif a == b or b == c or a == c:
        print("Tam giác cân")
    else:
        print("Tam giác thường")
else:
    print("Không phải tam giác")
```

W3A16     [Rounding] 
Nhập vào một số thực: in ra kết quả làm tròn lên, tròn xuống, làm tròn tới số 
nguyên gần nhất (không được sử dụng hàm sẵn có của pyton). Các kết quả cách nhau một dấu cách

```python
x = float(input())
# Làm tròn xuống
t = int(x)
if t > x:
    xuong = t - 1
else:
    xuong = t
# Làm tròn lên
if t < x:
    len = t + 1
else:
    len = t
# Làm tròn tới số nguyên gần nhất (0.5 làm tròn lên)
y = x + 0.5
s = int(y)
if s > y:
    gan_nhat = s - 1
else:
    gan_nhat = s
print(xuong, len, gan_nhat)
```

W3A17  Cấp số nhân 
Cho 4 số a, b, c, d. Hãy kiểm tra xem 4 số này có thể theo thứ tự tạo thành 1 cấp số nhân với công bội nguyên theo đúng thứ tự a, b, c, d hay không? 
Gợi ý : Tìm công bội (b / a) nếu b chia hết cho a, sau đó lấy b nhân công bội và so sánh vs c, c nhân công bội và so sánh vs d.

```python
a, b, c, d = map(int, input().split())
if a == 0:
    if b == 0 and c == 0 and d == 0:
        print("Yes")
    else:
        print("No")
elif b % a != 0:
    print("No")
else:
    q = b // a
    if b * q == c and c * q == d:
        print("Yes")
    else:
        print("No")
```
